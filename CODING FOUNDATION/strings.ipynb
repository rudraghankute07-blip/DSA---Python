```python
s = 'rhythm'
vowels = 0
Consonents = 0
for ch in s:
    if ch == " ":
        continue
    if ch in "aeiou":
        vowels += 1
    else:
        Consonents += 1
print("vowels= ",vowels , "consonenets = ", Consonents)
```

```python
# reverse a string
s = "hello"
print(s[::-1])

# manual
text = "data"
rev = ""

for char in text:
    rev = char + rev

print(rev)


# concept
print("ab" + "cd")
```

```python
# palindrome
s = 'A man a plan a canal Panama'
s = s.lower()
s = s.replace(" ","")

rev = ""

for ch in s:
    rev = ch + rev

if (rev == s):
    print("palindrome")
else:
    print("not a palindrome")
```

```python
# anagram
A='listen'
B='silent'
if (sorted(A) == sorted(B)):
    print("anagram")
else:
    print("not anagram")
```

```python
# count word frequency
S='to be or not to be'
W='be'
S = S.lower().split()
W = W.lower()
count= 0

for words in S:
    if (W == words):
        count += 1
print(count)
```
