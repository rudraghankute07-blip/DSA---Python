```python
# euclidian algoritm 
# brute force 
N1 = 20  
N2 = 40
gcd = 0

for i in range(1 , min(N1 , N2) + 1):
    if (N1 % i == 0 and N2 % i == 0):
        gcd = i

print(gcd)
```
