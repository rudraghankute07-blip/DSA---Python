```python
# Max, Min, Sum and Average of Array
arr=[3,7,1,9,4]

max = arr[0]
min = arr[0]
sum = 0

for i in range (len(arr)):
    if arr[i] > max:
        max = arr[i]
    if arr[i] < min:
        min = arr[i]
    sum += arr[i]

print(max)
print(min)
print(sum)
print(f"{sum/len(arr):.2f}")
```

```python
# frequency counter
arr=[1,2,2,3,3,3,4]
freq = {}

for num in arr:
    if num in freq:
        freq[num] += 1

    else:
        freq[num] = 1

print(dict(freq.items()))
```

```python
# Find All Primes in Array and Their Sum
arr=[2,3,4,5,6,7,8,9,10]
prime = []
sum = 0
for i in arr:
    if i <= 1:
        continue
    for j in range(2 , (int(i ** 0.5))+1 ):
        if (i % j == 0):
            break
    else:
        prime.append(i)
        sum += i
print(prime , sum)
```
