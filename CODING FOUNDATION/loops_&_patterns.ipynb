```python
# Sum of Odd Numbers in Range
a = 1
b = 10
sum = 0
for i in range(a , b +1 , 2):
    print(i)
    sum += i
print("sum = ", sum)

# using n % 2 == 1
c = 1
d = 10
add = 0
for i in range(c , d + 1):
    if (i % 2 == 1):
        print(i)
        add += i
print(add)
```

```python
# digital root
n = 12345

while n >= 10 :
    sum = 0

    while (n > 0):
        sum += n % 10
        n = n // 10
    n = sum
print(n)


# method 2 - formula using % 9
n = 12345

if n == 0:
    result = 0
else:
    result = n % 9

    if result == 0:
        result = 9
print(result)
```

```python
# Right-Aligned Star Triangle
for i in range(1 , 6):
    print(i * "*")

# nesxted loops 
for i in range(1 , 6):
    for j in range(i):
        print("*" , end="")
    print()
```

```python
for i in range (5 , 0 , -1):
    print(i * "*")


for i in range(5 , 0 , -1):
    for j in range(i):
        print("*",end="")
    print()
```

```python
# Number Triangle Pattern
num = 1
for i in range(1 , 6):
    for j in range(i):
        print(num, end=" ")
        num += 1
    print()
```

```python
n = 5

for i in range(1, n + 1):

    # spaces
    for j in range(n - i):
        print(" ", end="")

    # stars
    for j in range(2 * i - 1):
        print("*", end="")

    print()
```
