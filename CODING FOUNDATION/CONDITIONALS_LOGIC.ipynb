```python
# Simple Calculator with Error Handling
a = int(input("a"))
b = int(input("b"))
op = input("op")

match op:
    case '+' :
        print('a + b : ', a + b )
    case '-' :
        print(a - b)
    case 'x' :
        print( a * b)
    case '/' :
        if (b == 0):
            print("ZeroDivisionError")
        else:
            print( a / b)
    case '%' :
        if (b == 0):
            print("ZeroDivisionError")
        else:
            print( a % b)
    case _:
        print("invalid oprator") 
```

```python
# tempreture scale 
tempreture , scale = input().split()
tempreture = float(tempreture)


match scale:
    case 'c':
        print("F :",(tempreture * 9/5) + 32)
        print("K :", tempreture + 273.15)
    case 'f':
        print("C :", (tempreture - 32) * 5/9)
        print("K :", (tempreture - 32) * 5/9 + 273.15)
    case 'K':
        print("C :",tempreture - 273.15)
        print("F :",(tempreture - 273.15) * 9/5 + 32)
```

```python
# fizzBuzz
n = 15

for i in range(1 , n +1 ):
    if (i % 15 == 0):
        print("FizzBuzz")
    elif (i % 3 == 0):
        print("Fizz")
    elif(i % 5 ==0 ):
        print("Buzz")
    else:
        print(i)
```
