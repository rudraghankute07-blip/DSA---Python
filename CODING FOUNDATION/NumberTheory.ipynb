```python
# palindrome 
n = 121
N = n
digit = 0

while ( n > 0):
    rev = n % 10
    digit = digit * 10 + rev
    n = n // 10
print(digit)

print("palindrome" if digit == N else "not palindrome")
```

```python
n = 153
N = n
M = n
angstrong = 0
count = 0

while (n > 0):
    count += 1
    n = n // 10
print(count)

while(N > 0):
    last_digit = N % 10
    angstrong += (last_digit ** count)
    N = N // 10

if (angstrong == M):
    print("angstorng")
else:
    print("False")
```

```python
# perfect number
n = 28
count = 0
for i in range(1, n//2 +1):
    if (n % i == 0):
        count += i
if(count == n):
    print("perfect number")
else:
    print("not perfect")
```

```python
# number of divisors
n = 36
arr = []
for i in range(1 , int(n**0.5)+1 ):
    if (n % i == 0):
        arr.append(i)
        if (n//i in arr):
            continue
        else:
            arr.append(n // i)
print(arr)




# number of divisors
n = 36
arr = []
for i in range(1 , int(n**0.5)+1 ):
    if (n % i == 0):
        arr.append(i)
        if (i != n//6):
            arr.append(n // i)
print(arr)
```
